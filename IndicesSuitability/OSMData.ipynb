```python
import contextily as cx
import osmnx as ox
```

```python
groceries = ox.geometries_from_place("Richardson, TX", tags={"shop": ["grocery", "supermarket"]})
print(f"There are {len(groceries)} results in the groceries result")
groceries.head()
```

```python
ax = groceries.plot()
cx.add_basemap(ax=ax, crs=groceries.crs.to_string())
```

```python
cols2keep = ["geometry", "name", "shop", "brand"]
groceries.geometry = groceries.geometry.centroid
groceries[cols2keep].to_file("data/OSMGroceries.gpkg")
```
