```python
import matplotlib.pyplot as plt  # matplotlib for plotting
import numpy as np  # numpy for analyzing numbers
```

```python
income_data = df["Estimate!!Households!!Mean income (dollars)"]  # save the Pandas Series (like a list) to the income_data variable
income_data  # view the variable. Normally, we have to use the print function, but Jupyter will just show us if it is the last thing in a cell.
```

```python
plt.hist(income_data)  # pyplot has a "hist" function for creating histograms!
```

```python
x = [i/5.0 for i in range(100)]  # create a list of 0, 0.2, 0.4, ...
y = x + np.random.normal(0, 1, 100)  # an array of random variables with mean 0, std=1, and 100 elements.
plt.scatter(x, y)
plt.plot(x, x, color='red', alpha=0.5)
```
