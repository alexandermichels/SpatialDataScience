## Geopandas

Okay, so what does this have to do with GIS? Well, Python is a powerful tool for spatial analysis!

We will walk through the basics of the Geopandas package! [geopandas](https://geopandas.org/en/stable/index.html) is a geospatial analysis package that extends the popular pandas package. In this section, we will import some basic cartographic boundary files from the U.S. Census Bureau: https://www.census.gov/geographies/mapping-files/time-series/geo/cartographic-boundary.html

First, we need to unzip the data. We can do this in Python using the following code block:

```python
# download the files from the Census
# the ! allows us to run terminal/Bash commands!
# wget is a linux command to download data from a link
!wget https://www2.census.gov/geo/tiger/GENZ2023/shp/cb_2023_us_county_500k.zip
```

```python
import zipfile  # import the zipfile package (default package in Python)

# we can use a with statement to ensure that our zipfile is closed after our operation
with zipfile.ZipFile("cb_2023_us_county_500k.zip", 'r') as zip_ref:  # open the zipfile to read ('r') and label the opened file as "zip_ref"
    zip_ref.extractall("./")  # extract the zipfile to the current directory. In linux, "./" refers to the current directory
```

Now, we can load the shapefiles using Geopandas! Let's start with the US County data.

```python
import geopandas as gpd  # load it with a nickname so we don't need to spell out "geopandas" every time

gdf = gpd.read_file("cb_2023_us_county_500k.shp")  # geopandas can load the data using the "read_file" function
gdf.head()  # this shows us the first 5 entries in the GeoDataFrame (like an Excel sheet with a Geoemetry column)
```

Now, let's try to plot the data. Geopandas is convenient because we can plot a GeoDataFrame simply using ".plot()"

```python
gdf.plot()
```

We can evaluate the Coordinate Reference System using the ".crs" attribute.

```python
gdf.crs
```

Geopandas can also plot the data in an interactive plot using the ".explore()" function:

```python
gdf.explore()
```

Let's filter the data down to get the lower 48 U.S. states. One way to do this is by filtering by the FIPS codes:

```python
FIPS_CODES_TO_REMOVE = [
    "02", # Alaska
    "03", # American Samoa
    "07", # Canal Zone
    "11", # DC
    "14", # Guam
    "15", # Hawaii
    "43", # Puerto Rico
    "52", # Virgin Islands
    "60", # American Samoa
    "66", # Guam
    "69", # Northern Mariana Islands
    "72", # Puerto Rico
    "78", # Virgin Islands
]
```

```python
states = gdf[~gdf.STATEFP.isin(FIPS_CODES_TO_REMOVE)]  # this filter asks if each entry's STATEFP is in our list, then the "~" reverses to "not in"
print(f"There are {len(states)} entries in the dataframe.")  # this uses an "f-string" which allows us to put variables in the strings we print!
states.plot()
```

Let's check that we didn't miss anything:

```python
print(len(set(states.STATEFP)))
print(sorted(set(states.STATEFP)))
print(sorted(set(states.STATE_NAME)))
```

We can easily plot Chloropleth maps with Geopandas as well by simply specifying the column we want to plot!

```python
states.plot(column="ALAND", legend=True)
```
